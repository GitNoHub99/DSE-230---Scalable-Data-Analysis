### 5. Get the last 100 characters of text in each row of the data and replace each space with a comma - 1 point

<ol>
    <li type ="a">Get the last 100 characters of text in each row of the dataframe.</li>
    <li type ="a">Replace each space with a comma in the resulting text.</li>
    <li type ="a">Alias/Rename the resulting column name to <code>last100comma</code> in the new dataframe</li>
    <li type ="a">Useful functions - <p><a href="https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.functions.substring.html#pyspark.sql.functions.substring">substring</a>, <a href="https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.Column.alias.html">Column.alias</a>, <a href="https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.DataFrame.select.html">DataFrame.select</a>, <a href="https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/api/pyspark.sql.functions.regexp_replace.html">regexp_replace</a></p></li>
</ol>

#### **Expected output**: First 10 rows of the resulting dataframe with a single column name `last100comma`

NOTE:  Pass <code>truncate=False</code> as argument to <code>DataFrame.show()</code>

```python
# DO NOT DELETE THIS CELL
from pyspark.sql.functions import substring, regexp_replace

df2 = df.select(
    regexp_replace(
        substring("value", -100, 100),
        " ", ", "
    ).alias("last100comma")
)

df2.show(10, truncate=False)
```

### 6. Save results to HDFS - 0.5 points

NOTE: Spark uses a distributed memory system, and stores working data in fragments known as "partitions". This is advantageous when a Spark cluster spans multiple machines, as each machine will only require part of the working data to do its own job. By default, Spark will save each of these data partitions into a individual file to avoid I/O collisions. We want only one output file, so we'll need to fuse all the data into a single partition first. 

Your task: 
<ol>
    <li type="a">Coalesce the previous dataframe to one partition using <code>DataFrame.coalesce(1)</code>. This returns a 1-partition dataframe. This makes sure that all our results will end up in the same text file.</li>
    <li type = "a">Save the 1-partition dataframe to HDFS using the <code>DataFrame.write.text(&lt;path&gt;)</code> method to the root directory of the HDFS, i.e. <code>hdfs:///BookReviews_1M_last100comma.txt</code>.</li>
</ol>

#### **Expected output**: None

```python
# YOUR CODE HERE
df2.coalesce(1).write.text("hdfs:///BookReviews_1M_last100comma.txt")
```
